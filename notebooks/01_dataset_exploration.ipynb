# 01 — Dataset Exploration

This notebook explores the video datasets used for the
Context-Aware AR for Freezing of Gait project.

## Objectives

1. Inspect available videos.
2. Examine video resolution and frame rate.
3. Understand video duration.
4. Visualize representative frames.
5. Identify environmental situations relevant to the project.

The initial focus is on understanding the data rather than training a model.

```python
from pathlib import Path

import cv2
import matplotlib.pyplot as plt

from cafog.utils.video import video_properties

DATA_DIR = Path("../data/raw")

print("Data directory:", DATA_DIR.resolve())
```

```python
videos = list(DATA_DIR.rglob("*.mp4"))

print(f"Found {len(videos)} video files")

for video in videos[:20]:
    print(video)
```

```python
if videos:
    video_path = videos[0]

    properties = video_properties(video_path)

    print("Video:", video_path)
    print(f"Width: {properties['width']:.0f}")
    print(f"Height: {properties['height']:.0f}")
    print(f"FPS: {properties['fps']:.2f}")
    print(f"Frames: {properties['frame_count']:.0f}")
    print(
        f"Duration: "
        f"{properties['duration_seconds']:.2f} seconds"
    )
else:
    print("No MP4 videos found yet.")
```

```python
if videos:
    capture = cv2.VideoCapture(str(videos[0]))

    total_frames = int(
        capture.get(cv2.CAP_PROP_FRAME_COUNT)
    )

    middle_frame = total_frames // 2

    capture.set(
        cv2.CAP_PROP_POS_FRAMES,
        middle_frame,
    )

    success, frame = capture.read()
    capture.release()

    if success:
        frame_rgb = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB,
        )

        plt.figure(figsize=(12, 7))
        plt.imshow(frame_rgb)
        plt.axis("off")
        plt.title(videos[0].name)
        plt.show()
```
