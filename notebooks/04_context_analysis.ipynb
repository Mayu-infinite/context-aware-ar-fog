# 04 — Context Analysis

This notebook investigates the transition from raw scene information to
environmental context.

## Core idea

Object detection answers:

> What is present?

Context understanding attempts to answer:

> What does the current environment mean for walking?

The desired pipeline is:

Scene observations
↓
Scene State
↓
Context
↓
Recommended Action
↓
Cue

```python
from cafog.context.rules import ContextEngine
from cafog.context.state import SceneState
from cafog.policy.cue_policy import CuePolicy

context_engine = ContextEngine()
cue_policy = CuePolicy()

print("Context engine ready.")
print("Cue policy ready.")
```

```python
example_states = [
    SceneState(),

    SceneState(
        door_detected=True,
        door_width_m=0.60,
        door_distance_m=1.0,
    ),

    SceneState(
        obstacle_detected=True,
        obstacle_distance_m=0.8,
    ),
]

for state in example_states:
    context = context_engine.classify(state)
    cue = cue_policy.select(context)

    print("=" * 60)
    print("Context:", context.value)
    print("Action:", cue.action.value)
    print("Visual:", cue.visual_cue)
    print("Audio:", cue.audio_cue)
```

```python
state = SceneState(
    door_detected=True,
    door_width_m=0.60,
    door_distance_m=1.0,
)

context = context_engine.classify(state)
cue = cue_policy.select(context)

print("Scene:")
print(state)

print()
print("Context:")
print(context.value)

print()
print("Recommended action:")
print(cue.action.value)

print()
print("Visual cue:")
print(cue.visual_cue)

print()
print("Audio cue:")
print(cue.audio_cue)
```
