# 02 — Video Analysis

This notebook investigates the temporal structure of walking videos.

## Objectives

- Read video frame by frame.
- Examine frame rate.
- Sample frames.
- Visualize motion over time.
- Understand how much temporal information is required to identify
  environmental context.

A key project principle is that a single frame may not be sufficient to
determine context such as "approaching a narrow doorway".

```python
from pathlib import Path

import cv2
import matplotlib.pyplot as plt

from cafog.utils.video import iter_video_frames

DATA_DIR = Path("../data/raw")

videos = list(DATA_DIR.rglob("*.mp4"))

print(f"Videos found: {len(videos)}")
```

```python
if videos:
    video_path = videos[0]

    frames = []

    for frame_index, frame in iter_video_frames(
        video_path,
        stride=30,
    ):
        frames.append(
            (frame_index, frame)
        )

        if len(frames) >= 6:
            break

    print(
        f"Collected {len(frames)} sample frames."
    )
```

```python
if frames:
    fig, axes = plt.subplots(
        2,
        3,
        figsize=(15, 8),
    )

    for ax, (frame_index, frame) in zip(
        axes.flat,
        frames,
    ):
        frame_rgb = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB,
        )

        ax.imshow(frame_rgb)
        ax.set_title(
            f"Frame {frame_index}"
        )
        ax.axis("off")

    plt.tight_layout()
    plt.show()
```
